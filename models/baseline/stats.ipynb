```python
import matplotlib.pyplot as plt
import numpy as np

models  = ['Random Forest', 'Logistic Regression', 'GCN - A']
w8_acc  = [0.74,  0.73,  0.74]
w12_acc = [0.83,  0.81,  0.814]
w8_f1   = [0.7,  0.714, 0.73]   # GCN(a) W8 F1 pending
w12_f1  = [0.81,  0.79,  0.81]

x      = np.arange(len(models))
width  = 0.35
C_W8   = '#85B7EB'
C_W12  = '#185FA5'

fig, axes = plt.subplots(1, 2, figsize=(10, 4))

for ax, w8, w12, title in zip(
    axes,
    [w8_acc,  w8_f1],
    [w12_acc, w12_f1],
    ['Accuracy', 'F1 Score']
):
    b8  = ax.bar(x - width/2, w8,  width, label='Week 8',  color=C_W8,  zorder=3)
    b12 = ax.bar(x + width/2, w12, width, label='Week 12', color=C_W12, zorder=3)

    for bar in [*b8, *b12]:
        h = bar.get_height()
        if not np.isnan(h):
            ax.text(bar.get_x() + bar.get_width() / 2, h + 0.004,
                    f'{h:.3f}', ha='center', va='bottom', fontsize=8)

    ax.set_title(title, fontsize=12, pad=8)
    ax.set_xticks(x)
    ax.set_xticklabels(models, fontsize=10)
    ax.set_ylim(0.63, 0.91)
    ax.yaxis.grid(True, color='#e8e8e8', zorder=0)
    ax.set_axisbelow(True)
    ax.spines[['top', 'right', 'left']].set_visible(False)
    ax.tick_params(left=False)

axes[0].legend(frameon=False, fontsize=10, loc='upper right')
axes[1].legend(frameon=False, fontsize=10, loc='upper right')
fig.tight_layout(pad=2.0)
plt.savefig('model_comparison_bars.png', bbox_inches='tight')
plt.show()
```
